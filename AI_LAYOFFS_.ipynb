```python
top_companies = df[['company', 'jobs_cut']].sort_values(by='jobs_cut', ascending=False)
top_companies.head(10)
```

```python
top10 = top_companies.head(10)
plt.figure(figsize=(12,6))
sns.barplot(x='company', y='jobs_cut', data=top10)
plt.title("Top 10 Companies by Layoffs")
plt.xlabel("Company")
plt.ylabel("Jobs Cut")
plt.xticks(rotation=45)
plt.show()
```

```python
sector_layoffs = df.groupby('sector')['jobs_cut'].sum().sort_values(ascending=False)
sector_layoffs
```

```python
plt.figure(figsize=(14,7))
sector_layoffs.plot(kind='bar')
plt.title("Total Layoffs by Sector")
plt.xlabel("Sector")
plt.ylabel("Total Jobs Cut")
plt.xticks(rotation=60, ha='right')
plt.tight_layout()
plt.show()
```

```python
layoffs_over_time = df.groupby('layoff_date')['jobs_cut'].sum()
layoffs_over_time
```

```python
plt.figure(figsize=(14,6))
plt.plot(layoffs_over_time.index, layoffs_over_time.values)
plt.title("Layoff Trends Over Time")
plt.xlabel("Date")
plt.ylabel("Total Jobs Cut")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
